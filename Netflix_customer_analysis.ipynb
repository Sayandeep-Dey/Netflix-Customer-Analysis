```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df.replace(["NULL","N/A",""," "],np.nan,inplace=True)
```

```python
df=df.apply(lambda x:x.str.strip() if x.dtype=="object" else x)
```

```python
df["Subscription_Plan"].value_counts()
```

```python
df["Title"]=df["Title"].str.title()
df["Content_Type"]=df["Content_Type"].str.title()
df["Genre"]=df["Genre"].str.title()
df["Country"]=df["Country"].str.title()
df["Subscription_Plan"]=df["Subscription_Plan"].str.title()
df["Device"]=df["Device"].str.title()
df["Payment_Method"]=df["Payment_Method"].str.title()
df["Account_Status"]=df["Account_Status"].str.title()
df["Customer_Name"]=df["Customer_Name"].str.title()
```

```python
df["Subscription_Plan"].value_counts()
```

```python
df["Account_Status"].value_counts()
```

```python
cols = ["Watch_Hours","Monthly_Fee","Movies_Watched","Episodes_Watched","Avg_Rating","Monthly_Sessions","Download_Count","Age"]
for c in cols:
    df[c]=pd.to_numeric(df[c],errors="coerce")
```

```python
df.info()
```

```python
df["Join_Date"]=pd.to_datetime(df["Join_Date"],errors="coerce")
df["Last_Watch_Date"]=pd.to_datetime(df["Last_Watch_Date"],errors="coerce")
```

```python
df.info()
```

```python
df=df[(df["Age"]>=18) & (df["Age"]<=100)]
```

```python
df=df[df["Watch_Hours"]>=0]
df=df[df["Monthly_Fee"]>=0]
df=df[df["Movies_Watched"]>=0]
df=df[df["Episodes_Watched"]>=0]
df=df[df["Avg_Rating"]>=0]
df=df[df["Monthly_Sessions"]>=0]
df=df[df["Download_Count"]>=0]
```

```python
df["Watch_Hours"]=df["Watch_Hours"].fillna(df["Watch_Hours"].mean())
df["Age"]=df["Age"].fillna(df["Age"].mean())
df["Monthly_Fee"]=df["Monthly_Fee"].fillna(df["Monthly_Fee"].mean())
```

```python
df["Genre"]=df["Genre"].fillna("Unknown")
df["Device"]=df["Device"].fillna("Unknown")
df["Payment_Method"]=df["Payment_Method"].fillna("Unknown")
```

```python
df["Customer_Renevue"]=df["Monthly_Fee"]*df["Monthly_Sessions"]
```

```python
df["Engagement_Score"]=(df["Watch_Hours"]+df["Monthly_Sessions"]+df["Download_Count"])
```

```python
bins=[0,20,50,100,np.inf]
label=["Low","Medium","High","Very High"]
df["Watch_Category"]= pd.cut(df["Watch_Hours"],bins=bins,labels=label)
```

```python
df.head()
```

```python
bins=[17,25,35,50,100]
label=["18-25","26-35","36-50","51+"]
df["Age_Group"]=pd.cut(df["Age"],bins=bins,labels=label)
```

```python
df["Active_Flag"]=np.where(df["Account_Status"]=="Active",1,0)
```
